```python
# AUGMENTATION BLOCK 1: Calculate Current Baseline & Setup Progressive Training

print("\n🚀 PROGRESSIVE AUGMENTATION - BUILDING ON SUCCESS")
print("Current: Bias eliminated! Now optimizing for 75% accuracy target")

# 🎯 BASELINE PERFORMANCE CALCULATION
def calculate_baseline_performance():
    """Calculate detailed baseline metrics from our successful bias-free model"""
    
    print("\n📊 CALCULATING BASELINE PERFORMANCE...")
    
    # Check if model exists, if not create a placeholder or load it
    try:
        # Test set evaluation
        test_predictions = micro_model.predict(x_test, verbose=0)
    except NameError:
        print("⚠️ micro_model not found. Please ensure the model is loaded or trained first.")
        print("Creating placeholder metrics for demonstration...")
        # Return placeholder metrics
        return {
            'accuracy': 0.0,
            'f1_macro': 0.0,
            'precision_macro': 0.0,
            'recall_macro': 0.0,
            'bias_score': 0.0,
            'per_class_metrics': {
                'precision': [0.0, 0.0],
                'recall': [0.0, 0.0],
                'f1': [0.0, 0.0]
            },
            'class_distribution': [0.5, 0.5]
        }
    test_pred_classes = np.argmax(test_predictions, axis=1)
    test_true_classes = np.argmax(y_test_2class, axis=1)
    
    # Calculate comprehensive metrics
    from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
    
    accuracy = accuracy_score(test_true_classes, test_pred_classes)
    precision_macro = precision_score(test_true_classes, test_pred_classes, average='macro', zero_division=0)
    recall_macro = recall_score(test_true_classes, test_pred_classes, average='macro', zero_division=0)
    f1_macro = f1_score(test_true_classes, test_pred_classes, average='macro', zero_division=0)
    
    # Per-class metrics
    precision_per_class = precision_score(test_true_classes, test_pred_classes, average=None, zero_division=0)
    recall_per_class = recall_score(test_true_classes, test_pred_classes, average=None, zero_division=0)
    f1_per_class = f1_score(test_true_classes, test_pred_classes, average=None, zero_division=0)
    
    # Bias calculation
    class_0_count = np.sum(test_pred_classes == 0)
    class_1_count = np.sum(test_pred_classes == 1)
    total = len(test_pred_classes)
    bias_score = abs(class_0_count - class_1_count) / total
    
    print(f"📋 BASELINE PERFORMANCE (Bias-Free Model):")
    print(f"  Overall Accuracy: {accuracy:.3f} ({accuracy*100:.1f}%)")
    print(f"  Macro F1-Score: {f1_macro:.3f}")
    print(f"  Macro Precision: {precision_macro:.3f}")
    print(f"  Macro Recall: {recall_macro:.3f}")
    print(f"  Bias Score: {bias_score:.3f}")
    
    print(f"\n📋 Per-Class Baseline:")
    class_names = ['Bean Rust', 'Angular Leaf Spot']
    for i, name in enumerate(class_names):
        print(f"  {name}:")
        print(f"    Precision: {precision_per_class[i]:.3f}")
        print(f"    Recall: {recall_per_class[i]:.3f}")
        print(f"    F1-Score: {f1_per_class[i]:.3f}")
        print(f"    Predictions: {(class_0_count if i==0 else class_1_count)}/{total} ({(class_0_count if i==0 else class_1_count)/total:.1%})")
    
    return {
        'accuracy': accuracy,
        'f1_macro': f1_macro,
        'precision_macro': precision_macro,
        'recall_macro': recall_macro,
        'bias_score': bias_score,
        'per_class_metrics': {
            'precision': precision_per_class,
            'recall': recall_per_class,
            'f1': f1_per_class
        },
        'class_distribution': [class_0_count/total, class_1_count/total]
    }

# Store baseline performance
baseline_metrics = calculate_baseline_performance()

# 🎯 PROGRESSIVE AUGMENTATION STRATEGY
print(f"\n🎯 PROGRESSIVE AUGMENTATION STRATEGY:")
print(f"  Phase 1: Light augmentation (preserve bias elimination)")
print(f"  Phase 2: Medium augmentation (improve generalization)")  
print(f"  Phase 3: Advanced augmentation (maximize performance)")
print(f"  Target: 75% accuracy while maintaining <30% bias")

# 🎨 AUGMENTATION PIPELINES

# Import from the correct location
try:
    import tensorflow as tf
    from tensorflow.keras.preprocessing.image import ImageDataGenerator
except ImportError:
    # For newer TensorFlow versions, try alternative imports
    try:
        import tensorflow as tf
        from tensorflow.keras.utils import image_dataset_from_directory
        # Try to use tf.keras.preprocessing directly
        ImageDataGenerator = tf.keras.preprocessing.image.ImageDataGenerator
    except (ImportError, AttributeError):
        # If still not available, install keras-preprocessing
        %pip install keras-preprocessing==1.1.2
        from keras_preprocessing.image import ImageDataGenerator
        import tensorflow as tf

def create_augmentation_pipelines():
    """Create three progressive augmentation pipelines"""
    
    # Phase 1: Light augmentation (safe, preserve current success)
    light_augmentation = ImageDataGenerator(
        rotation_range=15,               # Conservative rotation
        width_shift_range=0.1,           # Small shifts
        height_shift_range=0.1,          # Small shifts
        horizontal_flip=True,            # Natural variation
        brightness_range=[0.9, 1.1],     # Minimal brightness change
        fill_mode='nearest'
    )
    
    # Phase 2: Medium augmentation (balanced improvement)
    medium_augmentation = ImageDataGenerator(
        rotation_range=20,               # More rotation
        width_shift_range=0.15,          # Medium shifts
        height_shift_range=0.15,         # Medium shifts
        shear_range=0.1,                 # Light shear
        zoom_range=0.15,                 # Light zoom
        horizontal_flip=True,            # Natural variation
        brightness_range=[0.8, 1.2],     # More brightness variation
        channel_shift_range=10,          # Color shifts
        fill_mode='nearest'
    )
    
    # Phase 3: Advanced augmentation (maximum diversity)
    advanced_augmentation = ImageDataGenerator(
        rotation_range=25,               # Full rotation range
        width_shift_range=0.2,           # Full shifts
        height_shift_range=0.2,          # Full shifts
        shear_range=0.15,                # Medium shear
        zoom_range=0.25,                 # Full zoom range
        horizontal_flip=True,            # Natural variation
        brightness_range=[0.7, 1.3],     # Full brightness range
        channel_shift_range=20,          # Full color shifts
        fill_mode='nearest'
    )
    
    print(f"\n🎨 Augmentation Pipelines Created:")
    print(f"  ✅ Light: Conservative (preserve success)")
    print(f"  ✅ Medium: Balanced (improve generalization)")
    print(f"  ✅ Advanced: Aggressive (maximize diversity)")
    
    return light_augmentation, medium_augmentation, advanced_augmentation

# Create augmentation pipelines
light_datagen, medium_datagen, advanced_datagen = create_augmentation_pipelines()

# 🔄 ENHANCED BALANCED GENERATOR WITH AUGMENTATION
class EnhancedBalancedGenerator:
    """Enhanced generator with augmentation while maintaining perfect balance"""
    
    def __init__(self, X, y, datagen, batch_size=8, verbose=False):
        self.X = X
        self.y = y
        self.datagen = datagen
        self.batch_size = batch_size
        self.verbose = verbose
        
        # Ensure even batch size for perfect 50/50 split
        if batch_size % 2 != 0:
            self.batch_size = batch_size + 1
            print(f"⚠️ Adjusted batch size to {self.batch_size} for perfect balance")
        
        self.samples_per_class = self.batch_size // 2
        
        # Get class indices
        self.class_0_indices = np.where(np.argmax(y, axis=1) == 0)[0]  # Bean Rust
        self.class_1_indices = np.where(np.argmax(y, axis=1) == 1)[0]  # Angular Leaf Spot
        
        print(f"🔄 Enhanced Balanced Generator:")
        print(f"  Batch size: {self.batch_size}")
        print(f"  Samples per class per batch: {self.samples_per_class}")
        print(f"  Bean Rust samples available: {len(self.class_0_indices)}")
        print(f"  Angular Leaf Spot samples available: {len(self.class_1_indices)}")
        
        # Fit the data generator
        self.datagen.fit(X)
        
    def generate(self):
        """Generate perfectly balanced augmented batches"""
        batch_count = 0
        
        while True:
            batch_count += 1
            
            # Sample exactly equal amounts from each class
            class_0_batch_indices = np.random.choice(self.class_0_indices, self.samples_per_class, replace=True)
            class_1_batch_indices = np.random.choice(self.class_1_indices, self.samples_per_class, replace=True)
            
            # Get original images and labels
            class_0_images = self.X[class_0_batch_indices]
            class_1_images = self.X[class_1_batch_indices]
            class_0_labels = self.y[class_0_batch_indices]
            class_1_labels = self.y[class_1_batch_indices]
            
            # Apply augmentation to each image individually
            augmented_batch_X = []
            augmented_batch_y = []
            
            # Augment Bean Rust samples
            for i in range(len(class_0_images)):
                img = class_0_images[i].reshape(1, 256, 256, 3)
                aug_iter = self.datagen.flow(img, batch_size=1)
                augmented_img = next(aug_iter)[0]
                augmented_batch_X.append(augmented_img)
                augmented_batch_y.append(class_0_labels[i])
            
            # Augment Angular Leaf Spot samples
            for i in range(len(class_1_images)):
                img = class_1_images[i].reshape(1, 256, 256, 3)
                aug_iter = self.datagen.flow(img, batch_size=1)
                augmented_img = next(aug_iter)[0]
                augmented_batch_X.append(augmented_img)
                augmented_batch_y.append(class_1_labels[i])
            
            # Convert to arrays
            batch_X = np.array(augmented_batch_X)
            batch_y = np.array(augmented_batch_y)
            
            # Shuffle to mix classes randomly
            shuffle_indices = np.random.permutation(len(batch_X))
            batch_X = batch_X[shuffle_indices]
            batch_y = batch_y[shuffle_indices]
            
            # Verify balance (debug check)
            if self.verbose and batch_count % 20 == 0:
                class_counts = np.sum(batch_y, axis=0)
                print(f"Batch {batch_count}: Bean Rust={class_counts[0]}, Angular Leaf Spot={class_counts[1]}")
            
            yield batch_X, batch_y

# 📊 PROGRESSIVE TRAINING MONITOR
class ProgressiveTrainingMonitor(tf.keras.callbacks.Callback):
    """Monitor that tracks both performance improvement and bias maintenance"""
    
    def __init__(self, X_val, y_val, phase_name, baseline_metrics):
        super().__init__()
        self.X_val = X_val
        self.y_val = y_val
        self.phase_name = phase_name
        self.baseline_metrics = baseline_metrics
        self.best_epoch = 0
        self.best_improvement = 0
        
    def on_epoch_end(self, epoch, logs=None):
        # Get current predictions
        predictions = self.model.predict(self.X_val, verbose=0)
        pred_classes = np.argmax(predictions, axis=1)
        true_classes = np.argmax(self.y_val, axis=1)
        
        # Calculate current metrics
        current_accuracy = np.mean(pred_classes == true_classes)
        current_f1 = f1_score(true_classes, pred_classes, average='macro', zero_division=0)
        
        # Calculate bias
        class_0_count = np.sum(pred_classes == 0)
        class_1_count = np.sum(pred_classes == 1)
        total = len(pred_classes)
        current_bias = abs(class_0_count - class_1_count) / total
        
        # Calculate improvement over baseline
        accuracy_improvement = current_accuracy - self.baseline_metrics['accuracy']
        f1_improvement = current_f1 - self.baseline_metrics['f1_macro']
        
        print(f"\n🔍 {self.phase_name} - Epoch {epoch + 1} Progress:")
        print(f"  📊 Current Performance:")
        print(f"    Accuracy: {current_accuracy:.3f} ({accuracy_improvement:+.3f} vs baseline)")
        print(f"    F1-Score: {current_f1:.3f} ({f1_improvement:+.3f} vs baseline)")
        print(f"    Bias: {current_bias:.3f}")
        
        print(f"  📋 Prediction Distribution:")
        print(f"    Bean Rust: {class_0_count}/{total} ({class_0_count/total:.1%})")
        print(f"    Angular Leaf Spot: {class_1_count}/{total} ({class_1_count/total:.1%})")
        
        # Track best improvement
        combined_improvement = accuracy_improvement + f1_improvement
        if combined_improvement > self.best_improvement:
            self.best_improvement = combined_improvement
            self.best_epoch = epoch + 1
            print(f"  🎯 NEW BEST! Combined improvement: {combined_improvement:+.3f}")
        
        # Status indicators
        if current_bias > 0.4:
            print(f"  🚨 HIGH BIAS WARNING! Consider stopping this phase.")
        elif current_bias < 0.3:
            print(f"  ✅ Bias under control")
        
        if accuracy_improvement > 0.05:
            print(f"  🎉 SIGNIFICANT ACCURACY IMPROVEMENT!")
        elif accuracy_improvement > 0:
            print(f"  📈 Accuracy improving steadily")

print(f"\n✅ PROGRESSIVE AUGMENTATION SETUP COMPLETE!")
print(f"Ready to execute 3-phase training for performance optimization!")
```
