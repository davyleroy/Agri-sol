```python
# Advanced data augmentation for imbalanced learning
def create_imbalanced_augmentation():
    """
    Create specialized augmentation for imbalanced bean disease detection
    Focuses on preserving disease characteristics while increasing diversity
    """
    
    # Conservative augmentation to preserve agricultural features
    train_datagen = ImageDataGenerator(
        rotation_range=15,           # Gentle rotation preserves leaf structure
        width_shift_range=0.1,       # Minimal shift to keep lesions visible
        height_shift_range=0.1,
        shear_range=0.1,            # Preserve disease morphology
        zoom_range=0.1,             # Maintain disease spot size ratios
        horizontal_flip=True,        # Natural for plant diseases
        vertical_flip=False,         # Unnatural for agricultural imagery
        brightness_range=[0.9, 1.1], # Preserve color disease indicators
        fill_mode='nearest',
        rescale=1./255
    )
    
    # Validation augmentation (minimal)
    val_datagen = ImageDataGenerator(rescale=1./255)
    
    return train_datagen, val_datagen

# Create augmentation generators
train_datagen, val_datagen = create_imbalanced_augmentation()

print("🎨 Imbalanced Learning Augmentation Strategy:")
print("✅ Disease-preserving transformations")
print("✅ Conservative parameters to maintain agricultural accuracy")
print("✅ Horizontal flip only (natural plant orientation)")
print("✅ Brightness preservation for color-based disease detection")

# Advanced SMOTE implementation for image data
def apply_smote_to_features(X, y, random_state=42):
    """
    Apply SMOTE to flattened image features for minority class oversampling
    """
    print("🔄 Applying SMOTE for minority class oversampling...")
    
    # Flatten images for SMOTE
    X_flattened = X.reshape(X.shape[0], -1)
    
    # Apply BorderlineSMOTE (more conservative than regular SMOTE)
    smote = BorderlineSMOTE(
        random_state=random_state,
        k_neighbors=3,  # Conservative for small minority classes
        sampling_strategy='auto'  # Automatically balance to majority class size
    )
    
    X_resampled, y_resampled = smote.fit_resample(X_flattened, y)
    
    # Reshape back to image format
    X_resampled = X_resampled.reshape(-1, IMG_HEIGHT, IMG_WIDTH, 3)
    
    print(f"✅ SMOTE completed:")
    print(f"   Original samples: {len(X)}")
    print(f"   Resampled samples: {len(X_resampled)}")
    
    # Show class distribution after SMOTE
    smote_dist = Counter(y_resampled)
    print(f"   Class distribution after SMOTE:")
    for i, class_name in enumerate(class_names):
        clean_name = class_name.replace('_', ' ').title()
        print(f"     {clean_name}: {smote_dist[i]} samples")
    
    return X_resampled, y_resampled

# Apply SMOTE to training data (optional - memory intensive)
APPLY_SMOTE = False  # Set to True if you have sufficient memory

if APPLY_SMOTE and len(X_train) < 5000:  # Only for manageable dataset sizes
    X_train_smote, y_train_smote = apply_smote_to_features(X_train, y_train)
    y_train_smote_cat = to_categorical(y_train_smote, NUM_CLASSES)
    print("✅ SMOTE applied to training data")
else:
    X_train_smote, y_train_smote_cat = X_train, y_train_cat
    print("⚠️  SMOTE skipped (using class weights instead for efficiency)")

print(f"\n🔧 Final Training Configuration:")
print(f"Training samples: {len(X_train_smote)}")
print(f"Validation samples: {len(X_val)}")
print(f"Test samples: {len(X_test)}")
print(f"Imbalance strategy: Class weights + Focal loss + Augmentation")
```
