```python
# Configure training parameters and callbacks
INITIAL_EPOCHS = 10    # Phase 1: Head training
FINE_TUNE_EPOCHS = 20  # Phase 2: Fine-tuning
TOTAL_EPOCHS = INITIAL_EPOCHS + FINE_TUNE_EPOCHS

# Calculate steps per epoch
STEPS_PER_EPOCH = train_generator.samples // BATCH_SIZE
VALIDATION_STEPS = validation_generator.samples // BATCH_SIZE

print(f"Training Configuration:")
print(f"✅ Steps per epoch: {STEPS_PER_EPOCH}")
print(f"✅ Validation steps: {VALIDATION_STEPS}")
print(f"✅ Total epochs: {TOTAL_EPOCHS}")

# Define callbacks for robust training
callbacks = [
    # Early stopping to prevent overfitting
    EarlyStopping(
        monitor='val_accuracy',      # Monitor validation accuracy
        patience=7,                  # Stop if no improvement for 7 epochs
        restore_best_weights=True,   # Restore best weights
        verbose=1,
        mode='max'
    ),
    
    # Model checkpoint to save best model
    ModelCheckpoint(
        filepath=str(MODEL_SAVE_PATH / 'agrisol_tomato_best_model.h5'),
        monitor='val_accuracy',
        save_best_only=True,
        save_weights_only=False,
        verbose=1,
        mode='max'
    ),
    
    # Reduce learning rate when loss plateaus
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,                  # Reduce LR by half
        patience=3,                  # Patience before reduction
        min_lr=1e-7,                # Minimum learning rate
        verbose=1
    )\n]\n\nprint(\"✅ Callbacks configured for optimal training\")\n\n# Compile model for Phase 1 (head training)\nmodel.compile(\n    optimizer=Adam(learning_rate=0.001),\n    loss='categorical_crossentropy',\n    metrics=['accuracy', 'top_3_accuracy']  # Include top-3 accuracy for multi-class\n)\n\nprint(\"✅ Model compiled for Phase 1 training (frozen backbone)\")\nprint(f\"✅ Trainable parameters: {sum([tf.reduce_prod(var.shape) for var in model.trainable_variables]):,}\")"
```

### Advanced Training Configuration
Our training strategy incorporates state-of-the-art techniques for optimal performance:

**1. Two-Phase Training Approach**
- **Phase 1 (Feature Learning)**: Frozen backbone, train classifier head (10 epochs)
- **Phase 2 (Fine-tuning)**: Unfrozen top layers, reduced learning rate (20 epochs)

```python
# Build the transfer learning model
def build_efficientnet_model(num_classes, input_shape=(224, 224, 3), dropout_rate=0.3):
    """
    Build EfficientNetB0-based transfer learning model
    
    Args:
        num_classes: Number of output classes
        input_shape: Input image dimensions
        dropout_rate: Dropout rate for regularization
    
    Returns:
        Compiled Keras model
    """
    # Load pre-trained EfficientNetB0 (without top classification layer)
    base_model = EfficientNetB0(
        weights='imagenet',          # Use ImageNet pre-trained weights
        include_top=False,           # Exclude final classification layer
        input_shape=input_shape,     # Specify input dimensions
        pooling='avg'                # Global average pooling
    )
    
    # Freeze base model layers initially for transfer learning
    base_model.trainable = False
    
    # Build custom classification head
    inputs = keras.Input(shape=input_shape, name='image_input')
    
    # Data augmentation layer (applied during training)
    x = tf.keras.utils.get_source_inputs(inputs)[0]  # Get raw input
    
    # EfficientNet feature extraction
    x = base_model(x, training=False)  # Set training=False for batch norm
    
    # Custom classification head
    x = layers.GlobalAveragePooling2D(name='global_avg_pooling')(x) if base_model.pooling != 'avg' else x
    x = layers.BatchNormalization(name='batch_norm')(x)
    x = layers.Dropout(dropout_rate, name='dropout_1')(x)
    
    # Dense layers for classification
    x = layers.Dense(512, activation='relu', name='dense_1')(x)
    x = layers.BatchNormalization(name='batch_norm_2')(x)
    x = layers.Dropout(dropout_rate/2, name='dropout_2')(x)
    
    x = layers.Dense(256, activation='relu', name='dense_2')(x)
    x = layers.Dropout(dropout_rate/2, name='dropout_3')(x)
    
    # Output layer with softmax activation for multi-class classification
    outputs = layers.Dense(num_classes, activation='softmax', name='predictions')(x)
    
    # Create the complete model
    model = Model(inputs, outputs, name='AgriSol_Tomato_EfficientNetB0')
    
    return model, base_model

# Build the model
model, base_model = build_efficientnet_model(NUM_CLASSES)

# Display model architecture
print("🏗️  Model Architecture Summary:")
print(f"✅ Base Model: EfficientNetB0 ({base_model.count_params():,} parameters)")
print(f"✅ Custom Head: {model.count_params() - base_model.count_params():,} parameters")
print(f"✅ Total Parameters: {model.count_params():,}")
print(f"✅ Trainable Parameters: {sum([tf.reduce_prod(var.shape) for var in model.trainable_variables]):,}")

# Visualize model architecture
model.summary()

# Plot model architecture
tf.keras.utils.plot_model(
    model, 
    to_file=str(MODEL_SAVE_PATH / 'tomato_model_architecture.png'),
    show_shapes=True,
    show_layer_names=True,
    rankdir='TB',
    dpi=150
)

print("✅ Model architecture diagram saved")
```
